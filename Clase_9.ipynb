# Argumentos tipo Tupla o *args

```python
def cuenta_final(*args):
    total_venta = sum(args)
    return total_venta
```

## Funciones con Argumentos tipo Diccionario o **kwargs

```python
def calificaciones(**kwargs):
    promedio = sum(kwargs.values()) / len(kwargs)
    num_alumnos = len(kwargs)
    return(num_alumnos, promedio)
```

```python
clase_406 = {"Juan": 10,
              "Pedro": 9,
                "Maria": 8,
                  "Luis": 7
                  }
```

```python
clase_406.keys()
```

```python
total_alumnos, promedio_clase = calificaciones(**clase_406)
```

```python
print(total_alumnos," ",promedio_clase)
```

## Combinando argumentos explicitos, default, *args y **kwargs

```python
def reporte_empresarial(area,*args,manager='CEO',**kwargs):
    print(f"Reporte de {manager} del área de {area}")
    print("Los empleados son: ")
    for empleado in args:
        print(empleado)
    print("Los resultados son: ")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
    return sum(kwargs.values())
```

```python
area = 'Marketing y Ventas'
empleados = ["Juan", "Pedro", "Maria", "Luis"]
resultados = {"Ventas": 10000, "Marketing": 5000, "Producción": 20000}
```

```python
reporte_empresarial(area, *empleados, **resultados)
```
