# Funciones con Argumentos posicionales o tipo tupla

```python
def promedio(alumno,*materias):
    suma = 0
    for materia in materias:
        suma += materia
    promedio = suma / len(materias)
    return alumno,promedio
```

```python
nombre, promedio_alumno =promedio("Esteban",10,10)
print(f"El promedio de {nombre} es: {promedio_alumno}")
```

```python
# Error en la posicion, deben ir primero los argumentos explicitos y después los tupla o pocicionales.
nombre, promedio_alumno =promedio(10,10,"Esteban")
print(f"El promedio de {nombre} es: {promedio_alumno}")
```

```python
nombre, promedio_alumno =promedio("Laura",7,5,10,8,6,5,2,1,9,9)
print(f"El promedio de {nombre} es: {promedio_alumno}")
```

# Combinando Argumentos explícitos, tupla y diccionario.

```python
def crear_personaje(nombre,especie, *valores_poderes, **características):
    poder_total = sum(valores_poderes)
    personaje = {
        "nombre": nombre,
        "especie": especie,
        "poder_total": poder_total,
        "características": características
    }
    return personaje
```

```python
mago = crear_personaje("Gandalf el gris", "Valar", 80, 80, 80, clase="Mago", nivel=90, raza="Valar")
```
