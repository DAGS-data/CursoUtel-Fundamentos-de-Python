## Try - Except

```python
try:
    precio = 10
    costo = 0
    relacion =  precio/costo
except:
    print("Ocurrió un error")
```

```python
try:
    precio = 10
    costo = 1
    relacion =  precio/costo

    dict = {"a": 1, "b": 2}
    value = dict["Llave_mala"]
except ValueError:
    print("Ocurrió un error numerico o de valor")
except ZeroDivisionError:
    print("Ocurrió un error de división por cero")
except IndexError:
    print("Ocurrió un error de índice")
except Exception as e:
    print("Ocurrió un error inesperado:", str(e))
```

```python
try: 
    archivo = open("archivo.txt", "r")
except FileNotFoundError:
    print("El archivo no se encontró")
else:
    contenido = archivo.read()
    print(contenido)
    archivo.close()
finally:
    print("Se ejecutó el bloque finally")
```
