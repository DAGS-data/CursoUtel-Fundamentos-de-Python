# Manejo de Archivos

### Función open()

```python
# r - read
archivo = open("ejemplo.txt","r",encoding="utf-8")
```

```python
archivo.read()
```

```python
archivo.close()
```

```python
# write 
archivo = open("ejemplo.txt","w",encoding="utf-8")
archivo.write("Hello Word")
```

```python
archivo.close()
```

```python
archivo = open("ejemplo.txt","w",encoding="utf-8")
archivo.write("UTF-8 (8-bit Unicode Transformation Format) es un formato de codificación de caracteres Unicode e ISO 10646 que utiliza símbolos de longitud variable. UTF-8 fue creado por Robert C. Pike y Kenneth L. Thompson. Está definido como estándar por la <RFC 3629> de la Internet Engineering Task Force (IETF).[1] Actualmente es una de las tres posibilidades de codificación reconocidas por Unicode y lenguajes web, o cuatro en ISO 10646.")
archivo.close()
```

```python
# append - a
archivo = open("ejemplo.txt","a",encoding="utf-8")
archivo.write("ESTE TEXTO SE AGREGA AL FINAL DEL ARCHIVO")
archivo.close()
```

# Patrón with

```python
with open("ejemplo.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.read()
    print(contenido)
```

```python
with open("ejemplo.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.readline()
    print(contenido)
```

```python
with open("ejemplo.txt","r",encoding="utf-8") as archivo:
    contenido = archivo.readlines()
    print(contenido)
```

## Formas de escritura con with

```python
with open("ejemplo.txt","w",encoding="utf-8") as archivo:
    archivo.write("Este texto es una sobreescritura con with.")
```

```python
with open("ejemplo.txt","w",encoding="utf-8") as archivo:
    archivo.write("""UTF-8 (8-bit Unicode Transformation Format) es un formato de codificación de caracteres
Unicode e ISO 10646 que utiliza símbolos de longitud variable. 
UTF-8 fue creado por Robert C. Pike y Kenneth L. Thompson.
Está definido como estándar por la <RFC 3629> de la Internet Engineering Task Force (IETF).
[1] Actualmente es una de las tres posibilidades de codificación reconocidas 
por Unicode y lenguajes web, o cuatro en ISO 10646.
ESTE TEXTO SE AGREGA AL FINAL DEL ARCHIVO""")
```

```python
with open("ejemplo.txt","a",encoding="utf-8") as archivo:
    archivo.write("\n\n Este texto se agrega al final del archivo.")
```

```python
# Escritura de un csv
with open("alumnos.csv","w",encoding="utf-8") as archivo:
    archivo.write("Nombre,Apellido,Edad\n")
    archivo.write("Diego,Galván,30\n")
    archivo.write("Juan,Pérez,25\n")
    archivo.write("María,Gómez,28\n")
```

```python
print("Hola este es un formulario de vacaciones")



continua = True
while continua:
    nombre = input("Ingrese su nombre: ")
    edad = input("Ingrese su edad: ")
    lugar = input("Ingrese su lugar de vacaciones: ")
    with open("formulario_vacaciones.csv","a",encoding="utf-8") as archivo:
        archivo.write("Nombre,Edad,Lugar de vacaciones\n") # Ecabezado del archivo
        archivo.write(f"{nombre},{edad},{lugar}\n") # Datos del usuario
    continua = input("¿Desea ingresar otro formulario? (s/n): ").lower() == "s"
    if continua == "s":
        pass
    else:
        print("Gracias por llenar el formulario. ¡Felices vacaciones!")
        continua = False
    
```
